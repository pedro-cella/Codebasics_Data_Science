# Data Validation with Pydantic

```python
class User:
    def __init__(self, user_id, name, age):
        self.id = user_id
        self.name = name
        self.age = age
```

```python
dhaval = User(10, "Dhaval", "30 years")
dhaval.age
```

```python
from pydantic import BaseModel
from typing import Optional

class User(BaseModel):
    id: int
    name: str
    age: int
    address: Optional[str] = None

dhaval = User(id=10, name="Dhaval", age=30, address="111 Patli Gali")
```

```python
dhaval.name
```

## Takeaways
- Pydantic is a **data validation library in Python** that uses Python type annotations to validate data, ensureing it conforms to expect formats and types.
- It **simplifies data** parsing by automatically converting incoming data to the appropriate data types, raising informative errors when data is invalid or missing.
- **Integration with other Python libraries and frameworks,** like FastAPI for web development, leverages Pydantic for request validation, streamlining API development and error handling.
